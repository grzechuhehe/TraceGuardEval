# Materiał badawczy TraceGuard GSN
Inspekcja zbioru, bez wywołań API i bez generowania wyników LLM.
Szkice referencji wymagają niezależnej weryfikacji człowieka.

```python
from pathlib import Path
from traceguard_core.dataset import load_dataset, validate_dataset
ROOT = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "pyproject.toml").is_file())
DATA = ROOT / "data/gsn-v2"
validate_dataset(DATA)
```

```python
from collections import Counter
dataset = load_dataset(DATA)
print("Podział:", Counter(e.split for e in dataset.manifest.entries))
print("Scenariusze:", len(dataset.scenarios))
print("Plan generatora, nie etykiety człowieka:", Counter(r["planned_class"] for r in dataset.drafts))
```

```python
trace = dataset.traces[0]
trace.judge_payload()
```

## Odtworzenie metryk po badaniu
Użyj `traceguard_core.analysis.report` z zapisanymi wynikami i zamrożoną referencją.
Nie obliczaj accuracy na planowanych klasach generatora. Instrukcja pełnego przebiegu znajduje się w README.

```python
from traceguard_core.analysis import report
references = ROOT / "artifacts/references.json"
results = ROOT / "artifacts/test-global.jsonl"
frozen = ROOT / "artifacts/frozen-global.json"
if all(p.exists() for p in [references, results, frozen]):
    summary = report(DATA, references, results, frozen)
    print(summary["metrics"])
    print(summary["cost"])
else:
    print("Brak zamrożonej referencji lub rzeczywistych wyników. Nie ma jeszcze metryk badania.")
```
